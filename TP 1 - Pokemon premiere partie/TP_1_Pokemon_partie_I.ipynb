## L'univers des Pokémons

Un Pokémon est un animal issu du monde des jeux vidéo. Chaque Pokémon possède des caractéristiques qui lui sont propres, à savoir son type (Pokémon d'herbe, Pokémon de feu ... ) lui donnant alors des facultés spéciales. Ainsi, le Pokémon nommé `Dracofeu`, de type `Feu` est capable de cracher du feu. De plus, on note l'existence de Pokémons légendaires très rares et très puissants (Pokémon diamant, Pokémon platine...).

Les Pokémons sont également de différentes générations faisant référence à l'évolution de l'univers des Pokémons. Les Pokémons de première génération sont ceux ayant pris naissance au tout début de la création de l'univers (création de 151 Pokémons) pour arriver à la huitième génération avec plus de 800 Pokémons (2020)!

Dans cet univers viennent ensuite les dresseurs de Pokémons. Les dresseurs les collectionnent et les élèvent dans le but de les faire combattre entre eux jusqu'au KO. Lors d'un combat de Pokémons, opposant deux dresseurs, l'un choisit l'animal qui sera le premier à entrer dans l'arène. En fonction de ce premier Pokémon, le second dresseur choisit un animal dans sa collection qui sera en mesure de battre le premier Pokémon. Si le premier Pokémon présent dans l'arène est de type feu, il y a de fortes chances que le Pokémon choisi par l'adversaire soit de type eau afin de contrer les attaques du premier.

Cela nécessite pour le dresseur d'avoir une bonne connaissance de chaque Pokémon pour choisir celui qui pourra devenir vainqueur du combat. Pour l'aider dans son choix, les caractéristiques des Pokémons sont répertoriées dans un document appelé Pokédex.

Calculer le nombre de combats menés par chaque Pokémon

```python
# Un Pokémon peut être en première position ou en seconde position
combats_first = df_combats['First_pokemon'].value_counts()
combats_second = df_combats['Second_pokemon'].value_counts()

# Concaténer les deux séries et les sommer pour obtenir le nombre total de combats menés
nb_combats_menes = pd.concat([combats_first, combats_second], axis=0).groupby(level=0).sum()

print("\nNombre de combats menés par chaque Pokémon:")
print(nb_combats_menes.head())
```

```python
nb_combats_gagnes = df_combats['Winner'].value_counts()

print("\nNombre de combats gagnés par chaque Pokémon:")
print(nb_combats_gagnes.head())
```

Faire l'assemblage des observations (Préparation pour l'agrégation)

```python
# Créer un DataFrame pour les statistiques de combat
df_stats_combats = pd.DataFrame({
    'Pokemon_ID': df_pokedex['NUMERO'].unique() 
}).set_index('Pokemon_ID')

# Fusionner les nombres de combats menés et gagnés
df_stats_combats['Nb_Combats_Menes'] = nb_combats_menes
df_stats_combats['Nb_Combats_Gagnes'] = nb_combats_gagnes

# Remplacer les NaN (pour les Pokémon qui n'ont peut-être pas combattu ou gagné) par 0
df_stats_combats.fillna(0, inplace=True)

# Convertir en entier, car ce sont des comptages
df_stats_combats['Nb_Combats_Menes'] = df_stats_combats['Nb_Combats_Menes'].astype(int)
df_stats_combats['Nb_Combats_Gagnes'] = df_stats_combats['Nb_Combats_Gagnes'].astype(int)


print("\nStatistiques des combats par Pokémon (pré-agrégation):")
print(df_stats_combats.head())
```

```python
pokemon_plus_victoires_id = df_stats_combats['Nb_Combats_Gagnes'].idxmax()
nb_victoires_max = df_stats_combats['Nb_Combats_Gagnes'].max()
pokemon_plus_victoires_name = df_pokedex[df_pokedex['NUMERO'] == pokemon_plus_victoires_id]['NOM'].iloc[0]

print(f"\nLe Pokémon qui a gagné le plus de combats est : {pokemon_plus_victoires_name} (ID: {pokemon_plus_victoires_id}) avec {nb_victoires_max} victoires.")
```

```python
pokemon_jamais_gagnes = df_stats_combats[df_stats_combats['Nb_Combats_Gagnes'] == 0]

print("\nPokémons n'ayant jamais gagné de combat:")
if not pokemon_jamais_gagnes.empty:
    # Pour afficher les noms, nous devons fusionner avec df_pokedex
    pokemon_jamais_gagnes_names = df_pokedex[df_pokedex['NUMERO'].isin(pokemon_jamais_gagnes.index)]['NOM']
    print(pokemon_jamais_gagnes_names.tolist())
    print(f"Total de {len(pokemon_jamais_gagnes)} Pokémons n'ayant jamais gagné.")
else:
    print("Tous les Pokémons ont gagné au moins un combat.")
```

PARTIE 03 - Agrégation des données avec le Pokédex

```python
print("\n--- Agrégation des données avec le Pokédex ---")

# Calcul du pourcentage de victoires
df_stats_combats['Pourcentage_Victoires'] = df_stats_combats.apply(
    lambda row: (row['Nb_Combats_Gagnes'] / row['Nb_Combats_Menes']) * 100 if row['Nb_Combats_Menes'] > 0 else 0,
    axis=1
)

# Réinitialiser l'index pour que 'Pokemon_ID' devienne une colonne normale
df_stats_combats.reset_index(inplace=True)
df_stats_combats.rename(columns={'Pokemon_ID': 'NUMERO'}, inplace=True) # Renommer pour la fusion

# Fusionner df_pokedex avec df_stats_combats
# Nous allons utiliser le 'NUMERO' (ID du Pokémon) comme clé de fusion
df_pokedex_agg = pd.merge(df_pokedex, df_stats_combats, on='NUMERO', how='left')

# Remplir les NaN résultant de la fusion (si un Pokémon du Pokedex n'a pas participé à un combat)
# Avec 0 pour les comptages et pourcentages
df_pokedex_agg['Nb_Combats_Menes'].fillna(0, inplace=True)
df_pokedex_agg['Nb_Combats_Gagnes'].fillna(0, inplace=True)
df_pokedex_agg['Pourcentage_Victoires'].fillna(0, inplace=True)

# Convertir en entier si nécessaire après remplissage
df_pokedex_agg['Nb_Combats_Menes'] = df_pokedex_agg['Nb_Combats_Menes'].astype(int)
df_pokedex_agg['Nb_Combats_Gagnes'] = df_pokedex_agg['Nb_Combats_Gagnes'].astype(int)

print("\nNouveau Pokédex agrégé avec les statistiques de combats (df_pokedex_agg):")
# Afficher les colonnes pertinentes pour vérifier
print(df_pokedex_agg[['NUMERO', 'NOM', 'Nb_Combats_Menes', 'Nb_Combats_Gagnes', 'Pourcentage_Victoires', 'LEGENDAIRE']].head())

print(f"\nDimensions du nouveau Pokédex agrégé : {df_pokedex_agg.shape}")
```
