```python
# Implementing max pooling and average pooling
import numpy as np

def max_pool2d(X, pool_size):
    (n_H_prev, n_W_prev) = X.shape
    (f_H, f_W) = pool_size
    n_H = n_H_prev // f_H
    n_W = n_W_prev // f_W
    Y = np.zeros((n_H, n_W))
    
    for h in range(n_H):
        for w in range(n_W):
            vert_start = h * f_H
            vert_end = vert_start + f_H
            horiz_start = w * f_W
            horiz_end = horiz_start + f_W
            
            Y[h, w] = np.max(X[vert_start:vert_end, horiz_start:horiz_end])
    
    return Y

def avg_pool2d(X, pool_size):
    (n_H_prev, n_W_prev) = X.shape
    (f_H, f_W) = pool_size
    n_H = n_H_prev // f_H
    n_W = n_W_prev // f_W
    Y = np.zeros((n_H, n_W))
    
    for h in range(n_H):
        for w in range(n_W):
            vert_start = h * f_H
            vert_end = vert_start + f_H
            horiz_start = w * f_W
            horiz_end = horiz_start + f_W
            
            Y[h, w] = np.mean(X[vert_start:vert_end, horiz_start:horiz_end])
    
    return Y

X = np.array([  [1, 2, 3, 0],
                [4, 5, 6, 1],
                [7, 8, 9, 2],
                [0, 1, 2, 3]    ])

print("Max Pooling Result:\n", max_pool2d(X, (2, 2)))
print("Average Pooling Result:\n", avg_pool2d(X, (2, 2)))
```

```python
# plot the results
import matplotlib.pyplot as plt
plt.subplot(1, 3, 1)
plt.title('Original')
plt.imshow(X, cmap='gray')
plt.subplot(1, 3, 2)
plt.title('Max Pooling')
plt.imshow(max_pool2d(X, (2, 2)), cmap='gray')
plt.subplot(1, 3, 3)
plt.title('Average Pooling')
plt.imshow(avg_pool2d(X, (2, 2)), cmap='gray')
plt.show()
```
